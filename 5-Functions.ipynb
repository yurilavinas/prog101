### Why use functions?
So far we coded our algorithms one line after the other.
- That's fine, but:
    1. Imagine what will happen in your code has 300 lines? What about 1000? (Check Linux Kernel https://en.wikipedia.org/wiki/Linux_kernel)

        1.1 Imagine explaining someone else what is happening on line 500.
        
        1.2 Imagine if someone else is explaining that to you!
    2. Imagine that you want to re-use some code you wrote. copy & paste?

That's why we use functions! We already used a few: *len()*, *print()*...
- These functions are so used, that they became part of the language (libraries, packages, extensions work the same way)
- We can create our own functions for the same reason: re-use!

Some bonuses: helps with organization, finding and fixing bugs/erros, coherence and facilitate understanding.

### Characteristics of functions
1. They have a name (what makes a good function name? Look online)
2. They have parameters, from 0 to any. (try to avoid using too many)
3. They have docstring (some text explaining what the function does)
4. They have the body, the main code
5. They have to return something 

### Functions structures
- def funName (parameter1, parameter2, ...):
    - """
    - input: parameter1, int; parameter2, float ...
    - Returns True, or a value, or...
    - use this part to briefly explain what the function does, what are its inputs and outputs, anything worth of attetion.
    - """
    - code
    - return  

```python
def is_even_with_return( i ):
    """ 
    Input: i, a positive int
    Returns True if i is even, otherwise False
    """
    remainder = i % 2
    return remainder == 0
```

```python
is_even_with_return(2)
```

```python
is_even_with_return(3)
```

# Variable scope

Look at this function again:

```python
def is_even_with_return( i ):
    """ 
    Input: i, a positive int
    Returns True if i is even, otherwise False
    """
    remainder = i % 2
    return remainder == 0
```

See that it uses *i* as the parameter and *remainder* as another variable? What happens if we already had used *i* or *remainder* outside of the variable?

```python
i = 10
is_even_with_return(5)
```

```python
remainder = None
print( is_even_with_return(5) )
print( remainder )
```

What we're seeing is the effect of what is called variable scope. In practice that says that the name of variable is unique on it's scope. Changing scopes allow us to use the same variable name multiple times, which is super good when using functions. Why?

Functions in python can have multiple parameters (arguments), even functions can be set as a parameter.
- That's right: a function calling another function!

```python
def func_a():
    print('inside func_a')

def func_b(y):
    print('inside func_b')
    return y
```

```python
func_a()
```

```python
5+func_b(2)
```

```python
def func_c(z):
    print('inside func_c:', z)
    return z()
```

```python
print(func_c(func_a))
```

The same idea applies to what a function can return. Almost anything! Check the code below, what does the fuction *f()* return? What's *x*?

```python
def f():
    def x(a, b):
        return a+b
    return x
    
# the first part, f(), returns a function object
# then apply that function with parameters 3 and 4
val = f()(3,4)
print(val)
```

# Local & global variables
Given what we saw about the scope of variable, changing a variable inside a function doesn't affect a variable that's outside of the function code, even when the variable has the same name. 

- All of these variables are called local variables:
    1. They only "work" inside their scope.

What would a global variable be?
