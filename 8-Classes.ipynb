# Classes

- Almost everything in Python is an object, with its properties and methods
- A Class is like an object constructor, or a "blueprint" for creating objects
    - a template for creating structures
- They hold both variables and functions into a single entity (a thing)


## How to create a simple class

1. Use the keyword class
2. Define variables and/or functions
3. Create an object (like we do for variables)
4. Access the variables/methods

```python
class MyClass:
  x = 5
```

```python
p1 = MyClass()
print(p1.x) 
```

## How to create a more interesting class

1. Set the main variables values using *__init__()*, (always executed when the class is being initiated, the first time)
2. Define variables and/or functions
3. Create an object (like we do for variables), with values (like we do for functions)
4. Access the variables/methods

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
    self.country = "Brazil"

  def myName(self):
    print("Hello my name is " + self.name)

prof = Person("Yuri", 33)

print(prof.name)
print(prof.age) 
print(prof.country) 
```

```python
prof.myName()
```

```python
newStudent = Person("John Doe", 99)
print(newStudent.name)
print(newStudent.age) 
print(newStudent.country) 
```

# What is the self thing?

- The self parameter is a reference to the current instance of the class.
    - It is used to access variables that belong to the class.
